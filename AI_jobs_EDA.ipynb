```python
df.nlargest(
    15, "ai_exposure_llm_gpt4"
)[[
    "occupation_title",
    "job_category",
    "median_annual_wage_usd",
    "ai_exposure_llm_gpt4"
]]
```

```python
df["ai_group"] = pd.qcut(
    df["ai_exposure_llm_gpt4"],
    2,
    labels=["Low AI Exposure", "High AI Exposure"]
)

df["growth_group"] = pd.qcut(
    df["projected_growth_pct_2024_2034"],
    2,
    labels=["Low Growth", "High Growth"]
)
```

```python
pd.crosstab(
    df["ai_group"],
    df["growth_group"]
)
```
